# Step 4 → Step 5: 재화 소모 행동 기반 사용자 세그먼트 분석

## 1. 분석 배경

1차 행동 루프 분석 결과, 재화 소모 단계까지 도달한 사용자 3,367명 중
358명이 결제로 전환되었으며, 해당 구간의 결제 전환율은 10.63%로 확인됨.

즉, 재화를 실제로 사용한 경험이 있는 사용자 중에서도
대부분은 결제로 이어지지 않았고 일부 사용자만 결제로 전환됨.

따라서 본 분석에서는 재화 소모 사용자 전체를 하나의 집단으로 보기보다,
사용자의 재화 사용 행동 특성에 따라 여러 집단으로 세분화하여
결제 전환과 관련된 행동 패턴을 확인하고자 함.

특히 재화를 얼마나 자주 사용하는지, 총 얼마나 많이 사용하는지,
한 번 사용할 때 얼마나 사용하는지 등 소비 행동의 차이를 기준으로
사용자군을 구분하고 각 집단의 결제 전환 특성을 비교하고자 함.

## 2. 분석 목적

Step 4 재화 소모 단계까지 도달한 사용자 3,367명을 대상으로,
재화 소모 행동 특성에 따라 사용자를 세분화하고
각 사용자 세그먼트의 결제 전환 특성을 비교하는 것을 목적으로 함.

단순히 결제자와 미결제자의 평균적인 차이를 비교하는 것에서 끝나지 않고,

- 어떤 재화 사용 행동에서 두 그룹의 차이가 나타나는지
- 어떤 행동 특성을 가진 사용자군에서 결제 전환율이 상대적으로 높거나 낮게 나타나는지
- 결제 전환과 관련성이 높은 사용자 행동 패턴을 구분할 수 있는지

를 확인하고자 함.

이를 통해 최종적으로는
모든 재화 소모 사용자를 동일하게 바라보는 것이 아니라,
행동 특성이 다른 사용자군을 구분하여
각 세그먼트에 적합한 결제 전환 전략을 제안할 수 있는 근거를 마련하고자 함.

## 3. 분석 질문

본 분석에서는 다음 질문을 중심으로 EDA와 세그먼트 분석을 진행함.

1. 결제자와 미결제자는 재화 소모 행동에서 어떤 차이를 보이는가?
2. 재화 소모 횟수, 총 소모량, 1회 평균 소모량 등 행동 변수 중
   어떤 특성에서 결제자와 미결제자의 차이가 크게 나타나는가?
3. 재화 소모 행동 변수들 사이에는 어떤 관계가 있으며,
   세그먼트 기준으로 함께 활용할 가치가 있는가?
4. 재화 소모 행동을 기준으로 의미 있는 사용자 세그먼트를 정의할 수 있는가?
5. 세그먼트별 결제 전환율은 어떻게 다른가?
6. 결제 전환율이 상대적으로 높은 세그먼트는 어떤 행동 특성을 보이는가?

## 4. 세그먼트 분석 방법

세그먼트 기준을 처음부터 임의로 결정하지 않을 것임.

먼저 재화 소모 단계까지 도달한 사용자들의 행동 데이터를 사용자 단위로 집계하고,
재화 소모 횟수, 총 소모량, 평균 소모량, 최대 소모량 등
재화 사용 특성을 나타내는 변수를 생성함.

이후 다음 순서로 분석을 진행함.

1. 사용자별 재화 소모 행동 변수 생성
2. 각 행동 변수의 분포 확인
3. 결제자와 미결제자의 행동 차이 비교
4. 행동 변수와 결제 전환율 간의 관계 탐색
5. 행동 변수 간 상관관계 및 중복성 확인
6. 세그먼트 기준으로 사용할 변수 선정
7. 선정된 행동 특성을 기준으로 사용자 세그먼트 생성
8. 세그먼트별 사용자 수와 결제 전환율 비교
9. 각 세그먼트의 행동 특성을 프로파일링하여 해석

세그먼트 생성에 사용할 변수와 구간 기준은
사전에 고정하지 않고 EDA 결과를 바탕으로 결정함.

또한 본 분석은 행동과 결제 전환 사이의 연관성을 탐색하는 것을 목적으로 하며,
특정 재화 소모 행동이 결제를 직접적으로 유발했다고 해석하지 않음.

### 분석 흐름

재화 소모 사용자 정의
→ 사용자 단위 행동 변수 생성
→ 결제자/미결제자 비교
→ 변수별 결제 전환 관계 확인
→ 변수 간 관계 확인
→ 세그먼트 기준 선정
→ 사용자 세그먼트 생성
→ 세그먼트별 결제 전환율 비교
→ 세그먼트 프로파일링 및 인사이트 도출

## 5. 분석 대상 정의

본 분석의 대상은 1차 행동 루프에서
재화 소모 단계(Step 4)까지 도달한 사용자 3,367명임.

해당 사용자 중 결제 이력이 있는 사용자를 결제자로,
결제 이력이 없는 사용자를 미결제자로 구분함.

- 전체 재화 소모 사용자: 3,367명
- 결제 사용자: 358명
- 미결제 사용자: 3,009명
- 결제 전환율: 10.63%

이후 모든 세그먼트 분석은 이 3,367명을 기준으로 진행함.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
```

```python
# 한글 폰트
plt.rcParams["font.family"] = "AppleGothic"
plt.rcParams["axes.unicode_minus"] = False
```

```python
# 프로젝트 경로
PROJECT_ROOT = Path.cwd().parents[3]
DATA_DIR = PROJECT_ROOT / "data" / "processed"

# 데이터 로드
friend_df = pd.read_csv(
    DATA_DIR / "accounts_friendrequest_preprocessed_2.csv"
)

question_df = pd.read_csv(
    DATA_DIR / "accounts_userquestionrecord_preprocessed_2.csv"
)

point_df = pd.read_csv(
    DATA_DIR / "accounts_pointhistory_preprocessed_2.csv"
)

payment_df = pd.read_csv(
    DATA_DIR / "accounts_paymenthistory_preprocessed_2.csv"
)

print("friend:", friend_df.shape)
print("question:", question_df.shape)
print("point:", point_df.shape)
print("payment:", payment_df.shape)
```

```python
# 친구 관계 형성 사용자
friend_users = set(
    pd.concat(
        [
            friend_df["send_user_id"],
            friend_df["receive_user_id"]
        ],
        ignore_index=True
    )
    .dropna()
    .unique()
)
```

```python
# 투표 수신 사용자
vote_received_users = set(
    question_df["chosen_user_id"]
    .dropna()
    .unique()
)
```

```python
# 친구 관계 형성 + 투표 수신
vote_stage_users = friend_users & vote_received_users
```

```python
# 재화 소모 사용자
spend_users = set(
    point_df["user_id"]
    .dropna()
    .unique()
)
```

```python
# Step 4 재화 소모 사용자
step4_users = vote_stage_users & spend_users

print("친구 관계 형성 사용자:", len(friend_users))
print("투표 수신 단계 사용자:", len(vote_stage_users))
print("Step 4 재화 소모 사용자:", len(step4_users))
```

```python
# 결제 사용자
payment_users = set(
    payment_df["user_id"]
    .dropna()
    .unique()
)
```

```python
# Step 4 사용자 중 결제자 / 미결제자
paid_users = step4_users & payment_users
non_paid_users = step4_users - payment_users

conversion_rate = len(paid_users) / len(step4_users) * 100

print("전체 분석 대상:", len(step4_users))
print("결제 사용자:", len(paid_users))
print("미결제 사용자:", len(non_paid_users))
print(f"결제 전환율: {conversion_rate:.2f}%")
```

## 6. 사용자 단위 행동 변수 생성

세그먼트 분석을 위해 재화 소모 로그를 사용자 단위로 집계함.

원본 `accounts_pointhistory` 데이터는 한 사용자가 재화를 사용할 때마다
여러 개의 행으로 기록되어 있기 때문에,
사용자별 행동 특성을 비교하려면 사용자 1명당 1행 형태로 변환할 필요가 있음.

본 분석에서는 다음 행동 변수를 생성함.

| 변수 | 의미 | 확인하고자 하는 행동 |
|---|---|---|
| `spend_count` | 재화 소모 횟수 | 얼마나 자주 재화를 사용했는가 |
| `total_spend` | 총 재화 소모량 | 누적해서 얼마나 많은 재화를 사용했는가 |
| `avg_spend` | 1회 평균 소모량 | 한 번 사용할 때 평균적으로 얼마나 사용했는가 |
| `max_spend` | 1회 최대 소모량 | 고액 재화 소모 경험이 있는가 |
| `spend_type_count` | 서로 다른 소모 금액 종류 수 | 얼마나 다양한 소모 패턴을 보였는가 |

이후 해당 변수들을 결제 여부와 함께 비교하여
세그먼트 기준으로 활용할 가치가 있는지 확인함.

```python
step4_point_df = point_df[
    point_df["user_id"].isin(step4_users)
].copy()

print("재화 소모 로그 수:", len(step4_point_df))
print("사용자 수:", step4_point_df["user_id"].nunique())
```

```python
# -30P 예외 기록 제거
step4_point_df = step4_point_df[
    step4_point_df["delta_point"] != -30
].copy()

print(step4_point_df["delta_point"].value_counts().sort_index())
```

```python
spend_user_df = (
    step4_point_df
    .groupby("user_id")
    .agg(
        spend_count=("delta_point", "count"),
        total_spend=("delta_point", lambda x: -x.sum()),
        avg_spend=("delta_point", lambda x: -x.mean()),
        max_spend=("delta_point", lambda x: -x.min()),
        spend_type_count=("delta_point", "nunique")
    )
    .reset_index()
)

spend_user_df.head()
```

```python
print("사용자 수:", spend_user_df["user_id"].nunique())
print("행 수:", len(spend_user_df))
```

```python
spend_user_df["paid"] = (
    spend_user_df["user_id"]
    .isin(paid_users)
    .astype(int)
)

spend_user_df["paid_label"] = spend_user_df["paid"].map({
    0: "미결제",
    1: "결제"
})
```

```python
print("전체 사용자:", len(spend_user_df))
print()
print(spend_user_df["paid_label"].value_counts())
```

## 7. 기초 EDA

사용자 단위로 생성한 재화 소모 행동 변수의 분포를 확인하고,
결제자와 미결제자 간 차이를 비교함.

본 단계의 목적은 각 변수가 결제 전환과 어떤 관계를 보이는지 탐색하고,
이후 사용자 세그먼트를 정의할 때 활용할 후보 변수를 선정하는 것임.

분석 대상 변수는 다음과 같음.

- `spend_count`: 재화 소모 횟수
- `total_spend`: 총 재화 소모량
- `avg_spend`: 1회 평균 재화 소모량
- `max_spend`: 1회 최대 재화 소모량
- `spend_type_count`: 서로 다른 소모 금액 종류 수

```python
behavior_cols = [
    "spend_count",
    "total_spend",
    "avg_spend",
    "max_spend",
    "spend_type_count"
]

spend_user_df[behavior_cols].describe().T.round(2)
```

```python
group_summary = (
    spend_user_df
    .groupby("paid_label")
    .agg(
        user_count=("user_id", "count"),

        spend_count_mean=("spend_count", "mean"),
        spend_count_median=("spend_count", "median"),

        total_spend_mean=("total_spend", "mean"),
        total_spend_median=("total_spend", "median"),

        avg_spend_mean=("avg_spend", "mean"),
        avg_spend_median=("avg_spend", "median"),

        max_spend_mean=("max_spend", "mean"),
        max_spend_median=("max_spend", "median"),

        spend_type_mean=("spend_type_count", "mean"),
        spend_type_median=("spend_type_count", "median")
    )
    .round(2)
)

group_summary
```

```python
behavior_cols = [
    "spend_count",
    "total_spend",
    "avg_spend",
    "max_spend",
    "spend_type_count"
]

median_compare = []

for col in behavior_cols:
    paid_median = spend_user_df.loc[
        spend_user_df["paid"] == 1, col
    ].median()

    non_paid_median = spend_user_df.loc[
        spend_user_df["paid"] == 0, col
    ].median()

    diff_rate = (
        (paid_median - non_paid_median)
        / non_paid_median
        * 100
    )

    median_compare.append({
        "variable": col,
        "paid_median": paid_median,
        "non_paid_median": non_paid_median,
        "difference_rate": diff_rate
    })

median_compare_df = pd.DataFrame(median_compare)

median_compare_df
```

```python
median_compare_df = median_compare_df.sort_values(
    "difference_rate",
    ascending=False
)

plt.figure(figsize=(9, 5))

bars = plt.bar(
    median_compare_df["variable"],
    median_compare_df["difference_rate"]
)

for bar, value in zip(
    bars,
    median_compare_df["difference_rate"]
):
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height() + 3,
        f"{value:.1f}%",
        ha="center"
    )

plt.axhline(0, linewidth=1)

plt.title("결제자와 미결제자의 행동 변수 중앙값 차이")
plt.xlabel("재화 소모 행동 변수")
plt.ylabel("미결제자 대비 결제자 중앙값 차이 (%)")

plt.show()
```

### 결제자와 미결제자 비교 결과

결제자와 미결제자의 재화 소모 행동을 비교한 결과,
가장 뚜렷한 차이는 재화 소모 횟수와 총 소모량에서 나타남.

- 재화 소모 횟수의 중앙값은 결제자가 29.5회,
  미결제자가 14회로 약 2.1배 높았음.
- 총 재화 소모량의 중앙값은 결제자가 6,375P,
  미결제자가 3,070P로 약 2.1배 높았음.
- 반면 1회 평균 소모량의 중앙값 차이는 약 1.3%로 매우 작았음.
- 최대 소모량의 중앙값은 두 그룹 모두 500P로 동일했음.
- 소모 금액 종류 수는 결제자가 다소 높았으나 차이는 상대적으로 작았음.

따라서 현재 단계에서는
`spend_count`와 `total_spend`를 주요 세그먼트 후보 변수로 봄.

다만 두 변수는 모두 재화 사용량을 나타내므로,
서로 유사한 정보를 포함하고 있는지 추가 확인할 필요가 있음.

### 주요 후보 변수 간 관계 확인

`spend_count`와 `total_spend`는 모두 결제자와 미결제자 간 차이가 크게 나타남.

하지만 두 변수가 매우 강하게 연관되어 있다면,
세그먼트 기준으로 동시에 사용하는 것이 중복될 수 있음.

따라서 산점도와 상관계수를 통해
두 변수 간 관계를 추가로 확인함.

```python
plt.figure(figsize=(7, 5))

plt.scatter(
    spend_user_df["spend_count"],
    spend_user_df["total_spend"],
    alpha=0.3
)

plt.title("재화 소모 횟수와 총 소모량의 관계")
plt.xlabel("재화 소모 횟수")
plt.ylabel("총 재화 소모량 (P)")

plt.show()
```

```python
corr_value = spend_user_df["spend_count"].corr(
    spend_user_df["total_spend"]
)

print(f"Pearson 상관계수: {corr_value:.3f}")
```

### 주요 후보 변수 간 관계 해석

`spend_count`와 `total_spend`의 Pearson 상관계수는 **0.594**로 나타남.

두 변수 사이에는 양의 상관관계가 존재하여,
재화 소모 횟수가 많은 사용자일수록 총 재화 소모량도 대체로 증가하는 경향이 확인됨.

다만 상관계수가 매우 높은 수준은 아니며,
산점도에서도 동일한 소모 횟수를 가진 사용자 사이에서
총 소모량의 차이가 비교적 크게 나타남.

이는 두 변수가 서로 관련되어 있으면서도 각각 다른 행동 특성을 포함하고 있음을 의미함.

- `spend_count`: 재화를 **얼마나 자주 반복적으로 사용했는지**를 나타내는 빈도 지표
- `total_spend`: 재화를 **누적해서 얼마나 많이 사용했는지**를 나타내는 규모 지표

따라서 현재 단계에서는 두 변수를 중복된 변수로 제거하지 않고,
사용자의 재화 소비 행동을 나타내는 주요 세그먼트 후보 변수로 유지함.

## 8. 사용자 세그먼트 생성

기초 EDA 결과, 재화 소모 행동 중
`spend_count`와 `total_spend`가 결제자와 미결제자 간
가장 큰 차이를 보이는 주요 행동 변수로 확인됨.

두 변수는 Pearson 상관계수 0.594로 양의 관계를 보였으나,
완전히 중복된 변수는 아닌 것으로 판단됨.

따라서 다음 두 행동 특성을 기준으로 사용자를 세분화함.

- `spend_count`: 재화 소비 빈도
- `total_spend`: 누적 재화 소비 규모

세그먼트 구간을 분석자가 임의로 설정하지 않고,
사용자 행동 데이터의 분포를 기반으로 군집을 생성하기 위해
K-Means 클러스터링을 활용함.

클러스터링 과정에서는 결제 여부(`paid`)를 사용하지 않으며,
재화 소비 행동만으로 세그먼트를 생성한 후
각 세그먼트의 결제 전환 특성을 비교함.

```python
segment_df = spend_user_df[
    ["user_id", "spend_count", "total_spend", "paid"]
].copy()

segment_df["log_spend_count"] = np.log1p(
    segment_df["spend_count"]
)

segment_df["log_total_spend"] = np.log1p(
    segment_df["total_spend"]
)

segment_df.head()
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_segment = scaler.fit_transform(
    segment_df[
        ["log_spend_count", "log_total_spend"]
    ]
)

X_segment[:5]
```

### 적절한 군집 수 탐색

K-Means의 군집 수를 임의로 결정하지 않고,
여러 군집 수에 대해 Elbow Method와 Silhouette Score를 확인하여
데이터 구조에 적절한 군집 수를 탐색함.

- Elbow Method: 군집 수를 늘렸을 때 군집 내부 오차가 감소하는 정도를 확인
- Silhouette Score: 같은 군집 내부의 응집도와 다른 군집과의 분리도를 함께 평가

두 결과와 각 군집의 해석 가능성을 종합하여 최종 군집 수를 결정함.

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

k_range = range(2, 7)

inertias = []
silhouette_scores = []

for k in k_range:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = kmeans.fit_predict(X_segment)

    inertias.append(kmeans.inertia_)
    silhouette_scores.append(
        silhouette_score(X_segment, labels)
    )

k_result = pd.DataFrame({
    "k": list(k_range),
    "inertia": inertias,
    "silhouette_score": silhouette_scores
})

k_result
```

```python
plt.figure(figsize=(7, 5))

plt.plot(
    k_result["k"],
    k_result["silhouette_score"],
    marker="o"
)

plt.title("군집 수에 따른 Silhouette Score")
plt.xlabel("군집 수 (k)")
plt.ylabel("Silhouette Score")
plt.xticks(k_result["k"])

plt.grid(alpha=0.3)
plt.show()
```

```python
plt.figure(figsize=(7, 5))

plt.plot(
    k_result["k"],
    k_result["inertia"],
    marker="o"
)

plt.title("군집 수에 따른 Inertia")
plt.xlabel("군집 수 (k)")
plt.ylabel("Inertia")
plt.xticks(k_result["k"])

plt.grid(alpha=0.3)
plt.show()
```

현재 두 변수만으로는 명확한 사용자 유형이 충분히 구분되지 않아, 재화 사용 방식의 차이를 반영할 수 있는 행동 변수를 추가하여 세그먼트 기준을 확장하기로 함.

## 8. 세그먼트 행동 변수 확장

초기 군집화에서는 `spend_count`와 `total_spend`를 활용했으나,
두 변수만으로는 명확한 사용자 유형이 충분히 구분되지 않았음.

두 변수 모두 사용자의 전반적인 재화 사용 강도를 나타내는 성격이 강하기 때문에,
단순한 사용량뿐 아니라 **재화를 어떤 방식으로 사용하는지**를 나타내는
행동 변수를 추가함.

추가 변수는 다음과 같음.

- `spend_active_days`: 재화를 사용한 서로 다른 날짜 수
- `spend_span_days`: 첫 재화 사용부터 마지막 재화 사용까지의 기간
- `spend_per_active_day`: 활동일 하루당 평균 재화 소모 횟수
- `minus10_ratio`: 전체 재화 소모 중 -10P 소모가 차지하는 비율

이를 통해 단기간 집중 사용, 장기간 반복 사용,
소액 반복 사용 등 기존 사용량 변수만으로는 구분하기 어려운
행동 패턴을 추가로 반영하고자 함.

```python
step4_point_df["created_at"] = pd.to_datetime(
    step4_point_df["created_at"]
)
```

```python
behavior_detail_df = (
    step4_point_df
    .groupby("user_id")
    .agg(
        spend_active_days=(
            "created_at",
            lambda x: x.dt.date.nunique()
        ),
        first_spend_at=(
            "created_at",
            "min"
        ),
        last_spend_at=(
            "created_at",
            "max"
        )
    )
    .reset_index()
)

behavior_detail_df["spend_span_days"] = (
    behavior_detail_df["last_spend_at"]
    - behavior_detail_df["first_spend_at"]
).dt.days

behavior_detail_df.head()
```

```python
# 사용자별 -10P 소모 횟수
minus10_df = (
    step4_point_df
    .assign(
        is_minus10=(
            step4_point_df["delta_point"] == -10
        ).astype(int)
    )
    .groupby("user_id", as_index=False)
    .agg(
        minus10_count=("is_minus10", "sum")
    )
)
```

```python
spend_user_df = (
    spend_user_df
    .merge(
        behavior_detail_df[
            [
                "user_id",
                "spend_active_days",
                "spend_span_days"
            ]
        ],
        on="user_id",
        how="left"
    )
    .merge(
        minus10_df,
        on="user_id",
        how="left"
    )
)
```

```python
spend_user_df["spend_per_active_day"] = (
    spend_user_df["spend_count"]
    / spend_user_df["spend_active_days"]
)

spend_user_df["minus10_ratio"] = (
    spend_user_df["minus10_count"]
    / spend_user_df["spend_count"]
)
```

```python
new_behavior_cols = [
    "spend_active_days",
    "spend_span_days",
    "spend_per_active_day",
    "minus10_ratio"
]

spend_user_df[new_behavior_cols].describe().T.round(2)
```

### 확장 행동 변수 분포 확인

추가한 행동 변수의 분포를 확인한 결과,
사용자마다 재화를 사용하는 방식에서 차이가 나타남.

- 재화 사용 활동일 수의 중앙값은 6일로 나타남.
- 첫 사용부터 마지막 사용까지의 기간 중앙값은 13일이었으며,
  일부 사용자는 장기간에 걸쳐 재화를 사용함.
- 활동일당 재화 소모 횟수의 중앙값은 2.5회로,
  재화를 사용할 때의 집중도에서도 사용자 간 차이가 존재함.
- `-10P` 소모 비율의 중앙값은 0.10이었으나,
  사용자별로 0~1까지 넓게 분포하여 소비 유형에서도 차이가 확인됨.

따라서 추가 변수들은 기존 `spend_count`, `total_spend`만으로는

행동 변수 간 상관관계를 확인한 결과,

- `total_spend`: 누적 재화 소비 규모
- `spend_per_active_day`: 활동일 기준 소비 집중도
- `spend_span_days`: 재화 사용 지속 기간

K-Means는 변수의 크기와 분포에 영향을 받기 때문에,
군집화 전에 각 변수의 분포를 다시 확인하고
필요한 경우 로그 변환 및 표준화를 적용함.

```python
cluster_features = [
    "log_total_spend",
    "log_spend_per_active_day",
    "log_spend_span_days",
    "minus10_ratio"
]
```

`total_spend`의 분포 변환 방법을 비교한 결과,

- 제곱근 변환 왜도: 0.38
- 로그 변환 왜도: -1.43

으로 나타남.

로그 변환은 분포의 치우침 방향만 반대로 변화시킨 반면,
제곱근 변환은 왜도의 절댓값을 가장 크게 감소시킴.

따라서 군집화에서는 `total_spend`에 제곱근 변환을 적용함.

```python
cluster_df["sqrt_total_spend"] = np.sqrt(
    cluster_df["total_spend"]
)

cluster_features = [
    "sqrt_total_spend",
    "log_spend_per_active_day",
    "log_spend_span_days",
    "minus10_ratio"
]
```

## 10. 결제 여부 차이 기반 사용자 세그먼트

비지도 군집화를 통해 자연적인 사용자 유형을 탐색하였으나,
Silhouette Score와 Elbow Method를 종합했을 때
명확하게 분리되는 군집 구조를 확인하기 어려웠음.

이에 따라 본 분석의 핵심 질문인

> "어떤 재화 소모 행동을 가진 사용자군에서 결제 비율의 차이가 나타나는가?"

에 보다 직접적으로 접근하기 위해
얕은 Decision Tree를 활용하여 행동 기반 세그먼트를 구성함.

Decision Tree는 결제 여부에 따라 차이가 나타나는 행동 변수와
분기 기준을 데이터에서 탐색할 수 있으며,
최종 Leaf Node를 각각 하나의 사용자 세그먼트로 해석함.

본 분석은 전체 관측기간의 행동을 활용한 탐색적 분석이므로,
세그먼트와 결제 간 관계는 연관성으로 해석하며
인과관계 또는 사전 결제 예측으로 해석하지 않음.

```python
segment_features = [
    "spend_count",
    "total_spend",
    "spend_span_days",
    "spend_per_active_day",
    "minus10_ratio"
]

X = spend_user_df[segment_features]
y = spend_user_df["paid"]
```

```python
# 각 사용자가 속한 Leaf ID
spend_user_df["behavior_leaf"] = behavior_tree.apply(X_behavior)

# Leaf별 사용자 구성
leaf_summary = (
    spend_user_df
    .groupby("behavior_leaf")
    .agg(
        user_count=("user_id", "count"),
        paid_users=("paid", "sum")
    )
    .reset_index()
)

leaf_summary["non_paid_users"] = (
    leaf_summary["user_count"]
    - leaf_summary["paid_users"]
)

leaf_summary["paid_rate"] = (
    leaf_summary["paid_users"]
    / leaf_summary["user_count"]
    * 100
)

leaf_summary["non_paid_rate"] = (
    leaf_summary["non_paid_users"]
    / leaf_summary["user_count"]
    * 100
)

# 결제율 낮은 순서 → 높은 순서로 정렬
leaf_summary = (
    leaf_summary
    .sort_values("paid_rate")
    .reset_index(drop=True)
)

# 중립적인 ID 부여
leaf_summary["group"] = [
    f"G{i+1}"
    for i in range(len(leaf_summary))
]

leaf_summary[
    [
        "group",
        "behavior_leaf",
        "user_count",
        "paid_users",
        "non_paid_users",
        "paid_rate",
        "non_paid_rate"
    ]
]
```
