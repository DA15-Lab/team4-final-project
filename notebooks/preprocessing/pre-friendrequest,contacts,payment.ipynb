#### 데이터 불러오기

```python
import pandas as pd

base_path = r"C:\Users\wjdtm\OneDrive\바탕 화면\CODEIT DA\Codeit DA Part.4 (2026.08.03 ~ 2026.10.02)\데이터 분석 고급 프로젝트\DATA_CSV"

# 10만 건씩 나누어서 읽은 뒤 리스트에 담아 합치기
chunk_size = 100000
chunks = []

for chunk in pd.read_csv(f"{base_path}\\accounts_friendrequest.csv", sep='\t', encoding='utf-8', chunksize=chunk_size):
    chunks.append(chunk)

df_friendrequest = pd.concat(chunks, ignore_index=True)

# 나머지 파일들은 용량이 작다면 그대로 로드
df_contacts = pd.read_csv(f"{base_path}\\accounts_user_contacts.csv", encoding='utf-8')
df_payment = pd.read_csv(f"{base_path}\\accounts_paymenthistory.csv", encoding='utf-8')
```

```python
# 날짜 컬럼 'created_at', 'updated_at' 데이터 타입 datetime으로 변경, id 컬럼 'id', 'user_id' 데이터 타입 object로 변경
df_friendrequest['created_at'] = pd.to_datetime(df_friendrequest['created_at'], errors='coerce')
df_friendrequest['updated_at'] = pd.to_datetime(df_friendrequest['updated_at'], errors='coerce')
df_friendrequest['id'] = df_friendrequest['id'].astype(object)
df_friendrequest['user_id'] = df_friendrequest['send_user_id'].astype(object)

# 변환 후 정보 확인
df_friendrequest.info()
```
