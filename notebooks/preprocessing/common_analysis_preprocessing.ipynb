```python
import pandas as pd
from pathlib import Path

PROJECT_ROOT = Path.cwd().parents[1]
```

```python
file_path = (
    PROJECT_ROOT
    / "data"
    / "interim"
    / "accounts_userquestionrecord_preprocessed.csv"
)

question_df = pd.read_csv(file_path)

print("원본 shape:", question_df.shape)
question_df.head()
```

```python
question_df["created_at"] = pd.to_datetime(
    question_df["created_at"],
    errors="coerce"
)

start_date = pd.Timestamp("2023-05-13")
end_date = pd.Timestamp("2024-05-08")  # 5/7 23:59:59까지 포함하기 위해

question_period = question_df[
    (question_df["created_at"] >= start_date) &
    (question_df["created_at"] < end_date)
].copy()

print("기간 필터 전:", question_df.shape)
print("기간 필터 후:", question_period.shape)

print(
    question_period["created_at"].min(),
    "~",
    question_period["created_at"].max()
)
```

```python
question_analysis = question_period[
    question_period["chosen_user_id"].notna()
].copy()

print("최종 shape:", question_analysis.shape)
print(
    "선택받은 고유 사용자 수:",
    question_analysis["chosen_user_id"].nunique()
)
```

```python
output_path = (
    PROJECT_ROOT
    / "data"
    / "interim"
    / "accounts_userquestionrecord_preprocessed_2.csv"
)

question_analysis.to_csv(output_path, index=False)

print("저장 완료:", output_path)
```

```python
file_path = (
    PROJECT_ROOT
    / "data"
    / "interim"
    / "accounts_pointhistory_preprocessed.csv"
)

point_df = pd.read_csv(file_path)

print("원본 shape:", point_df.shape)
point_df.head()
```

```python
point_df["created_at"] = pd.to_datetime(
    point_df["created_at"],
    errors="coerce"
)

start_date = pd.Timestamp("2023-05-13")
end_date = pd.Timestamp("2024-05-08")

point_analysis = point_df[
    (point_df["created_at"] >= start_date) &
    (point_df["created_at"] < end_date) &
    (point_df["delta_point"] < 0)
].copy()

print("필터 전:", point_df.shape)
print("필터 후:", point_analysis.shape)

print(
    point_analysis["created_at"].min(),
    "~",
    point_analysis["created_at"].max()
)

print(
    "포인트 소모 사용자 수:",
    point_analysis["user_id"].nunique()
)
```

```python
output_path = (
    PROJECT_ROOT
    / "data"
    / "interim"
    / "accounts_pointhistory_preprocessed_2.csv"
)

point_analysis.to_csv(output_path, index=False)

print("저장 완료:", output_path)
print("최종 shape:", point_analysis.shape)
```
